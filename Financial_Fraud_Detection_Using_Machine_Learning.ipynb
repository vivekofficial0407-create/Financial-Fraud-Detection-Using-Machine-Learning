```python
# To import the pandas library.
# It is used to load, manipulate and analyze tabular data.

import pandas as pd

# To import the NumPy library.
# It is used for numerical operations and working with arrays.

import numpy as np

# To import Matplotlib for data visualization.
# It is used to create different types of graphs and charts.

import matplotlib.pyplot as plt

# To import Seaborn for advanced data visualization.
# It is used to create attractive statistical graphs.

import seaborn as sns

```

```python
# To read the CSV dataset into a DataFrame.
# It loads the dataset from the specified file path.

df = pd.read_csv("/content/creditcard.csv")

# To display the first five rows of the dataset.
# It helps us understand how the data looks.

df.head()
```

```python
# To count the number of transactions in each class.
# It calculates how many Genuine and Fraud transactions are present.

class_count = df["Class"].value_counts()

# To create a figure with a specific size.
# It provides enough space to display the graph clearly.

plt.figure(figsize=(6,4))

# To create a bar plot for transaction classes.
# It compares the number of Genuine and Fraud transactions.

class_count.plot(kind="bar")

# To add a title to the graph.
# It describes the purpose of the graph.

plt.title("Fraud vs Genuine Transactions")

# To label the X-axis.
# It represents the transaction class.

plt.xlabel("Class (0 = Genuine, 1 = Fraud)")

# To label the Y-axis.
# It represents the number of transactions.

plt.ylabel("Number of Transactions")

# To display the graph.
# It renders the visualization.

plt.show()
```

The dataset contains significantly more Genuine transactions than Fraud transactions. This shows that the dataset is imbalance

```python
plt.figure(figsize=(8,4))

# To create a boxplot of transaction amounts.
# It helps identify outliers in the Amount column.

plt.boxplot(df["Amount"], vert=False)

# To add a title to the graph.
# It describes the graph.

plt.title("Boxplot of Transaction Amount")

# To label the X-axis.
# It represents transaction amount.

plt.xlabel("Transaction Amount")

# To display the graph.
# It renders the visualization.

plt.show()
```

```python
plt.figure(figsize=(16,12))

# To calculate the correlation between all numerical columns.
# It measures how strongly the features are related to each other.

correlation_matrix = df.corr()

# To create a correlation heatmap.
# It displays correlation values using different colors.

sns.heatmap(correlation_matrix, cmap="coolwarm")

# To add a title to the heatmap.
# It describes the graph.

plt.title("Correlation Heatmap")

# To display the graph.
# It renders the visualization.

plt.show()
```

Most features have weak correlation because the V1 to V28 features are already PCA-transformed. Some features show positive or negative correlation with the target variable.

```python
# To calculate the first quartile (Q1) of the Amount column.
# It represents the 25th percentile of the transaction amounts.

Q1 = df["Amount"].quantile(0.25)

# To calculate the third quartile (Q3) of the Amount column.
# It represents the 75th percentile of the transaction amounts.

Q3 = df["Amount"].quantile(0.75)

# To calculate the Interquartile Range (IQR).
# It measures the spread of the middle 50% of the data.

IQR = Q3 - Q1

# To calculate the lower boundary for detecting outliers.
# Values below this limit are considered outliers.

lower_bound = Q1 - 1.5 * IQR

# To calculate the upper boundary for detecting outliers.
# Values above this limit are considered outliers.

upper_bound = Q3 + 1.5 * IQR

# To display the lower and upper boundaries.
# It helps us understand the range of valid transaction amounts.

print("Lower Bound :", lower_bound)
print("Upper Bound :", upper_bound)
```

```python
# To create a new dataset after removing outliers.
# It stores the outlier-free data without changing the original dataset.

df_outlier = df[(df["Amount"] >= lower_bound) & (df["Amount"] <= upper_bound)]

# To display the updated dataset size.
# It confirms how many records remain after removing outliers.

df.shape
```

```python
plt.figure(figsize=(8,4))

# To create a boxplot after removing outliers.
# It helps verify whether most outliers have been removed.

plt.boxplot(df["Amount"], vert=False)

# To add a title to the graph.
# It describes the graph.

plt.title("Boxplot After Removing Outliers")

# To label the X-axis.
# It represents transaction amount.

plt.xlabel("Transaction Amount")

# To display the graph.
# It renders the visualization.

plt.show()
```

```python
# To separate all input features from the target variable.
# It stores all independent variables in X by removing the Class column.

X = df.drop("Class", axis=1)

# To separate the target variable.
# It stores the dependent variable (Fraud or Genuine) in y.

y = df["Class"]

# To display the first five rows of the input features.
# It helps verify that the features have been separated correctly.

X.head()
```

```python
# To display the first five values of the target variable.
# It helps verify the output labels.

y.head()
```

```python
# To import the train_test_split function.
# It is used to divide the dataset into training and testing sets.

from sklearn.model_selection import train_test_split

# To split the dataset into training and testing data.
# It reserves 20% data for testing and uses 80% data for training.

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# To display the shapes of the training and testing datasets.
# It helps verify that the data has been split correctly.

print("X_train :", X_train.shape)
print("X_test :", X_test.shape)
print("y_train :", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# To import the StandardScaler class.
# It is used to standardize the feature values before training the model.

from sklearn.preprocessing import StandardScaler

# To create an object of the StandardScaler class.
# It prepares the scaler for feature standardization.

scaler = StandardScaler()

# To learn the scaling parameters from the training data.
# It calculates the mean and standard deviation of each feature.

scaler.fit(X_train)

# To standardize the training data.
# It transforms the training features to have mean 0 and standard deviation 1.

X_train = scaler.transform(X_train)

# To standardize the testing data using the same scaling parameters.
# It transforms the testing features without recalculating the mean and standard deviation.

X_test = scaler.transform(X_test)

# To display the first five rows of the scaled training data.
# It helps verify that the scaling has been applied successfully.

X_train[:5]
```

```python
# To import the LogisticRegression model.
# It is used to build a classification model for predicting fraud transactions.

from sklearn.linear_model import LogisticRegression

# To create an object of the LogisticRegression model.
# It initializes the classification model.

lr = LogisticRegression(random_state=42)

# To train the Logistic Regression model.
# It learns the relationship between the input features and the target variable.

lr.fit(X_train, y_train)

# To make predictions on the testing data.
# It predicts whether each transaction is Genuine or Fraud.

y_pred_lr = lr.predict(X_test)

# To display the first ten predicted values.
# It helps verify the prediction output.

y_pred_lr[:10]
```

```python
# To import evaluation metrics.
# It is used to measure the performance of the classification model.

from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

# To calculate the accuracy of the model.
# It returns the percentage of correctly predicted transactions.

accuracy = accuracy_score(y_test, y_pred_lr)

print("Accuracy :", accuracy)
```

```python
# To generate the confusion matrix.
# It shows the number of correct and incorrect predictions.

cm = confusion_matrix(y_test, y_pred_lr)

print(cm)
```

```python
# To generate the classification report.
# It displays Precision, Recall, F1-Score and Support for each class.

print(classification_report(y_test, y_pred_lr))
```

The Logistic Regression model achieved very high overall accuracy. However, since the dataset is imbalanced, Precision, Recall and F1-Score were also analyzed. The model correctly identified most genuine transactions and detected a good proportion of fraudulent transactions.

```python
# To import the DecisionTreeClassifier model.
# It is used to build a tree-based classification model.

from sklearn.tree import DecisionTreeClassifier
```

```python
# To create an object of the DecisionTreeClassifier.
# It initializes the Decision Tree model using the Gini Index.

dt = DecisionTreeClassifier(criterion="gini", random_state=42)
```

```python
# To train the Decision Tree model.
# It learns the relationship between the input features and the target variable.

dt.fit(X_train, y_train)
```

```python
# To make predictions on the testing data.
# It predicts whether each transaction is Genuine or Fraud.

y_pred_dt = dt.predict(X_test)

# To display the first ten predicted values.
# It helps verify the prediction output.

y_pred_dt[:10]
```

```python
# To calculate the accuracy of the Decision Tree model.
# It returns the percentage of correctly predicted transactions.

dt_accuracy = accuracy_score(y_test, y_pred_dt)

print("Decision Tree Accuracy :", dt_accuracy)
```

```python
# To generate the confusion matrix.
# It shows the number of correct and incorrect predictions made by the Decision Tree model.

dt_cm = confusion_matrix(y_test, y_pred_dt)

print(dt_cm)
```

```python
# To generate the classification report.
# It displays Precision, Recall, F1-Score and Support for each class.

print(classification_report(y_test, y_pred_dt))
```

```python
# To import the RandomForestClassifier model.
# It is used to build an ensemble classification model using multiple decision trees.

from sklearn.ensemble import RandomForestClassifier
```

```python
# To create an object of the RandomForestClassifier.
# It initializes the Random Forest model with 100 decision trees.

rf = RandomForestClassifier(n_estimators=100, random_state=42)
```

```python
# To train the Random Forest model.
# It learns the relationship between the input features and the target variable.

rf.fit(X_train, y_train)
```

```python
# To make predictions on the testing data.
# It predicts whether each transaction is Genuine or Fraud.

y_pred_rf = rf.predict(X_test)

# To display the first ten predicted values.
# It helps verify the prediction output.

y_pred_rf[:10]
```

```python
# To calculate the accuracy of the Random Forest model.
# It returns the percentage of correctly predicted transactions.

rf_accuracy = accuracy_score(y_test, y_pred_rf)

print("Random Forest Accuracy :", rf_accuracy)
```

```python
# To generate the confusion matrix.
# It shows the number of correct and incorrect predictions made by the Random Forest model.

rf_cm = confusion_matrix(y_test, y_pred_rf)

print(rf_cm)
```

```python
# To generate the classification report.
# It displays Precision, Recall, F1-Score and Support for each class.

print(classification_report(y_test, y_pred_rf))
```

Among the three machine learning models, Random Forest achieved the best overall performance. It produced the highest accuracy (99.96%) along with the highest Precision (97%), Recall (88%) and F1-Score (93%) for fraud detection. Therefore, Random Forest was selected as the final model for this project.
